```python
def extract_url_features(urls):
    if not urls:
        return {
            "url_count": 0,
            "url_total_length": 0,
            "url_max_length": 0,
            "url_digit_count": 0,
            "url_dot_count": 0,
            "url_hyphen_count": 0,
            "url_slash_count": 0,
            "url_has_https": 0,
            "url_has_ip": 0,
            "url_has_at": 0,
            "url_is_shortened": 0
        }

    urls = [str(url) for url in urls]

    total_length = sum(len(url) for url in urls)
    max_length = max(len(url) for url in urls)

    digit_count = sum(
        sum(char.isdigit() for char in url)
        for url in urls
    )

    dot_count = sum(url.count(".") for url in urls)

    hyphen_count = sum(
        url.count("-")
        for url in urls
    )

    slash_count = sum(
        url.count("/")
        for url in urls
    )

    has_https = int(
        any(url.lower().startswith("https://") for url in urls)
    )

    ip_pattern = re.compile(
        r"(?:\d{1,3}\.){3}\d{1,3}"
    )

    has_ip = int(
        any(ip_pattern.search(url) for url in urls)
    )

    has_at = int(
        any("@" in url for url in urls)
    )

    shortened_domains = [
        "bit.ly",
        "tinyurl.com",
        "t.co",
        "cutt.ly",
        "goo.gl",
        "is.gd",
        "ow.ly",
        "shorturl.at"
    ]

    is_shortened = int(
        any(
            domain in url.lower()
            for url in urls
            for domain in shortened_domains
        )
    )

    return {
        "url_count": len(urls),
        "url_total_length": total_length,
        "url_max_length": max_length,
        "url_digit_count": digit_count,
        "url_dot_count": dot_count,
        "url_hyphen_count": hyphen_count,
        "url_slash_count": slash_count,
        "url_has_https": has_https,
        "url_has_ip": has_ip,
        "url_has_at": has_at,
        "url_is_shortened": is_shortened
    }
```

# **Number/digit features**

```python
def extract_number_features(text):
    text = str(text)

    digits = re.findall(r"\d", text)
    numbers = re.findall(r"\d+(?:[.,]\d+)?", text)

    return {
        "digit_count": len(digits),
        "number_count": len(numbers),
        "has_digits": int(len(digits) > 0)
    }
```

```python
number_features = df["text_clean"].apply(
    extract_number_features
)

number_features_df = pd.DataFrame(
    number_features.tolist(),
    index=df.index
)

df = pd.concat(
    [df, number_features_df],
    axis=1
)

print("✅ Number features added")
```

# **Text-length features**

```python
def extract_text_length_features(text):
    text = str(text).strip()

    words = text.split()

    return {
        "char_count": len(text),
        "word_count": len(words)
    }
```

```python
length_features = df["text_clean"].apply(
    extract_text_length_features
)

length_features_df = pd.DataFrame(
    length_features.tolist(),
    index=df.index
)

df = pd.concat(
    [df, length_features_df],
    axis=1
)

print("✅ Text-length features added")
```

# **Punctuation features**

```python
def extract_punctuation_features(text):
    text = str(text)

    return {
        "exclamation_count": text.count("!"),
        "question_count": text.count("?"),
        "comma_count": text.count(","),
        "dot_count": text.count("."),
        "colon_count": text.count(":"),
        "special_char_count": len(
            re.findall(r"[^\w\s]", text, flags=re.UNICODE)
        )
    }
```

# **Currency features**

```python
CURRENCY_PATTERN = re.compile(
    r"(?:৳|টাকা|BDT|Tk|USD|\$|€|£|¥)",
    re.IGNORECASE
)

def extract_currency_features(text):
    text = str(text)

    matches = CURRENCY_PATTERN.findall(text)

    return {
        "currency_count": len(matches),
        "has_currency": int(len(matches) > 0)
    }
```

```python
def extract_security_features(text):
    text = str(text).lower()

    found_keywords = []

    for keyword in SECURITY_KEYWORDS:
        if keyword.lower() in text:
            found_keywords.append(keyword)

    return {
        "security_keyword_count": len(found_keywords),
        "has_security_keyword": int(
            len(found_keywords) > 0
        )
    }
```

# **Bangla/English ratio**

```python
def extract_language_features(text):
    text = str(text)

    bangla_chars = re.findall(
        r"[\u0980-\u09FF]",
        text
    )

    english_chars = re.findall(
        r"[A-Za-z]",
        text
    )

    total_language_chars = (
        len(bangla_chars) +
        len(english_chars)
    )

    if total_language_chars == 0:
        bangla_ratio = 0.0
        english_ratio = 0.0
    else:
        bangla_ratio = (
            len(bangla_chars) /
            total_language_chars
        )

        english_ratio = (
            len(english_chars) /
            total_language_chars
        )

    return {
        "bangla_char_count": len(bangla_chars),
        "english_char_count": len(english_chars),
        "bangla_ratio": bangla_ratio,
        "english_ratio": english_ratio,
        "has_bangla": int(len(bangla_chars) > 0),
        "has_english": int(len(english_chars) > 0)
    }
```

```python
language_features = df["text_clean"].apply(
    extract_language_features
)

language_features_df = pd.DataFrame(
    language_features.tolist(),
    index=df.index
)

df = pd.concat(
    [df, language_features_df],
    axis=1
)

print("✅ Language features added")
```

```python
# ============================================================
# CELL 8 — Token-length analysis
# ============================================================

from tqdm.auto import tqdm

def get_token_lengths(texts, tokenizer, batch_size=64):
    lengths = []

    texts = list(texts)

    for start in tqdm(
        range(0, len(texts), batch_size),
        desc="Analyzing token lengths"
    ):
        batch = texts[start:start + batch_size]

        encoded = tokenizer(
            batch,
            add_special_tokens=True,
            padding=False,
            truncation=False,
            return_length=True
        )

        lengths.extend(encoded["length"])

    return np.array(lengths)


all_lengths = get_token_lengths(
    df["text_for_model"].astype(str).tolist(),
    tokenizer
)

print("Number of SMS:", len(all_lengths))

print("\nToken length statistics:")
print(
    pd.Series(all_lengths).describe(
        percentiles=[
            0.50,
            0.75,
            0.90,
            0.95,
            0.99
        ]
    )
)
```

```python
# ============================================================
# CELL 15 — PyTorch Dataset
# ============================================================

import torch
from torch.utils.data import Dataset


class BanglaSmishingDataset(Dataset):

    def __init__(
        self,
        encodings,
        structured_features,
        labels
    ):

        self.input_ids = torch.tensor(
            encodings["input_ids"],
            dtype=torch.long
        )

        self.attention_mask = torch.tensor(
            encodings["attention_mask"],
            dtype=torch.long
        )

        self.structured_features = torch.tensor(
            structured_features,
            dtype=torch.float32
        )

        self.labels = torch.tensor(
            labels,
            dtype=torch.long
        )

        # ----------------------------------------------------
        # Safety checks
        # ----------------------------------------------------

        assert len(self.input_ids) == len(
            self.attention_mask
        )

        assert len(self.input_ids) == len(
            self.structured_features
        )

        assert len(self.input_ids) == len(
            self.labels
        )

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):

        return {
            "input_ids": self.input_ids[idx],
            "attention_mask": self.attention_mask[idx],
            "structured_features": self.structured_features[idx],
            "labels": self.labels[idx]
        }
```

```python
# ============================================================
# 9. DRIVE FILE vs GITHUB FILE HASH CHECK
# ============================================================

import hashlib

def sha256_file(path):
    h = hashlib.sha256()

    with open(path, "rb") as f:
        while True:
            chunk = f.read(1024 * 1024)

            if not chunk:
                break

            h.update(chunk)

    return h.hexdigest()


drive_to_repo = {
    "raw_dataset":
        (
            os.path.join(
                PROJECT_DIR,
                "dataset",
                "hf_bangla_smishing.csv"
            ),
            os.path.join(
                REPO_DIR,
                "data",
                "raw",
                "hf_bangla_smishing.csv"
            )
        ),

    "train_processed":
        (
            os.path.join(
                PROJECT_DIR,
                "dataset",
                "train_processed.csv"
            ),
            os.path.join(
                REPO_DIR,
                "data",
                "processed",
                "train_processed.csv"
            )
        ),

    "val_processed":
        (
            os.path.join(
                PROJECT_DIR,
                "dataset",
                "val_processed.csv"
            ),
            os.path.join(
                REPO_DIR,
                "data",
                "processed",
                "val_processed.csv"
            )
        ),

    "test_processed":
        (
            os.path.join(
                PROJECT_DIR,
                "dataset",
                "test_processed.csv"
            ),
            os.path.join(
                REPO_DIR,
                "data",
                "processed",
                "test_processed.csv"
            )
        ),

    "scaler":
        (
            os.path.join(
                PROJECT_DIR,
                "models",
                "structured_feature_scaler.pkl"
            ),
            os.path.join(
                REPO_DIR,
                "artifacts",
                "structured_feature_scaler.pkl"
            )
        ),

    "preprocessing_config":
        (
            os.path.join(
                PROJECT_DIR,
                "results",
                "preprocessing_config.json"
            ),
            os.path.join(
                REPO_DIR,
                "preprocessing",
                "preprocessing_config.json"
            )
        ),

    "class_weights":
        (
            os.path.join(
                PROJECT_DIR,
                "results",
                "class_weights.json"
            ),
            os.path.join(
                REPO_DIR,
                "preprocessing",
                "class_weights.json"
            )
        ),

    "checkpoint":
        (
            os.path.join(
                PROJECT_DIR,
                "results",
                "preprocessing_checkpoint.json"
            ),
            os.path.join(
                REPO_DIR,
                "preprocessing",
                "preprocessing_checkpoint.json"
            )
        )
}

print("=" * 70)
print("GOOGLE DRIVE vs LOCAL GITHUB COPY")
print("=" * 70)

for name, (drive_path, github_path) in drive_to_repo.items():

    if not os.path.exists(drive_path):
        print(f"⚠️ {name}: Drive file missing")
        continue

    if not os.path.exists(github_path):
        print(f"❌ {name}: GitHub copy missing")
        continue

    drive_hash = sha256_file(drive_path)
    github_hash = sha256_file(github_path)

    if drive_hash == github_hash:
        print(f"✅ {name}: IDENTICAL")
    else:
        print(f"❌ {name}: DIFFERENT")
```

```python
# ============================================================
# FINAL PROJECT VERIFICATION
# Google Drive vs Local Repo vs Actual GitHub
# ============================================================

import os
import subprocess
import hashlib

PROJECT_DIR = "/content/drive/MyDrive/Bangla_Smishing_Project"
REPO_DIR = "/content/Multiclass-Bangla-Smishing-Detection"

# ------------------------------------------------------------
# 1. Make sure GitHub is up to date
# ------------------------------------------------------------

%cd "$REPO_DIR"

!git fetch origin main

# ------------------------------------------------------------
# 2. Get files from LOCAL repo
# ------------------------------------------------------------

local_files = []

for root, dirs, files in os.walk(REPO_DIR):
    dirs[:] = [d for d in dirs if d != ".git"]

    for file in files:
        full_path = os.path.join(root, file)

        relative = os.path.relpath(
            full_path,
            REPO_DIR
        ).replace("\\", "/")

        local_files.append(relative)

# ------------------------------------------------------------
# 3. Get files ACTUALLY on GitHub
# ------------------------------------------------------------

github_output = subprocess.check_output(
    [
        "git",
        "ls-tree",
        "-r",
        "--name-only",
        "origin/main"
    ],
    text=True
)

github_files = [
    x.strip()
    for x in github_output.splitlines()
    if x.strip()
]

local_set = set(local_files)
github_set = set(github_files)

# ------------------------------------------------------------
# 4. Print actual GitHub files
# ------------------------------------------------------------

print("=" * 70)
print("FILES ACTUALLY ON GITHUB")
print("=" * 70)

for f in sorted(github_files):
    print("✅", f)

# ------------------------------------------------------------
# 5. Files in local repo but NOT on GitHub
# ------------------------------------------------------------

only_local = sorted(
    local_set - github_set
)

print("\n" + "=" * 70)
print("LOCAL BUT NOT ON GITHUB")
print("=" * 70)

if only_local:
    for f in only_local:
        print("❌", f)
else:
    print("✅ None")

# ------------------------------------------------------------
# 6. Files on GitHub but NOT locally
# ------------------------------------------------------------

only_github = sorted(
    github_set - local_set
)

print("\n" + "=" * 70)
print("GITHUB BUT NOT LOCAL")
print("=" * 70)

if only_github:
    for f in only_github:
        print("⚠️", f)
else:
    print("✅ None")

# ------------------------------------------------------------
# 7. Important files check
# ------------------------------------------------------------

important_files = [
    "README.md",
    "requirements.txt",
    ".gitignore",

    "notebooks/Bangla_Smishing_Preprocessing.ipynb",

    "data/raw/hf_bangla_smishing.csv",

    "data/processed/train_processed.csv",
    "data/processed/val_processed.csv",
    "data/processed/test_processed.csv",

    "preprocessing/class_weights.json",
    "preprocessing/preprocessing_config.json",
    "preprocessing/preprocessing_checkpoint.json",

    "artifacts/structured_feature_scaler.pkl",

    "docs/Bangla_Smishing_Project_Handoff.md"
]

print("\n" + "=" * 70)
print("IMPORTANT FILES")
print("=" * 70)

all_important_ok = True

for f in important_files:

    exists_local = f in local_set
    exists_github = f in github_set

    if exists_local and exists_github:
        print("✅", f)

    elif exists_local:
        print("⚠️ LOCAL ONLY:", f)
        all_important_ok = False

    elif exists_github:
        print("⚠️ GITHUB ONLY:", f)
        all_important_ok = False

    else:
        print("❌ MISSING:", f)
        all_important_ok = False

# ------------------------------------------------------------
# 8. Commit synchronization
# ------------------------------------------------------------

local_commit = subprocess.check_output(
    ["git", "rev-parse", "HEAD"],
    text=True
).strip()

remote_commit = subprocess.check_output(
    ["git", "rev-parse", "origin/main"],
    text=True
).strip()

print("\n" + "=" * 70)
print("COMMIT SYNCHRONIZATION")
print("=" * 70)

print("Local commit :", local_commit)
print("GitHub commit:", remote_commit)

if local_commit == remote_commit:
    print("✅ Local repo and GitHub are synchronized.")
else:
    print("⚠️ Local repo and GitHub are NOT synchronized.")

# ------------------------------------------------------------
# 9. Git working tree
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("GIT STATUS")
print("=" * 70)

status = subprocess.run(
    ["git", "status", "--porcelain"],
    capture_output=True,
    text=True
)

if status.stdout.strip():
    print(status.stdout)
else:
    print("✅ Working tree clean.")

# ------------------------------------------------------------
# 10. FINAL RESULT
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FINAL RESULT")
print("=" * 70)

if (
    not only_local
    and
    all_important_ok
    and
    local_commit == remote_commit
    and
    not status.stdout.strip()
):
    print("🎉 EVERYTHING IS UPLOADED AND SYNCHRONIZED!")
else:
    print("⚠️ SOMETHING STILL NEEDS ATTENTION.")
```
